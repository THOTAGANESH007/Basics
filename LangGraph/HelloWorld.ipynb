```python
from typing import Dict, TypedDict
from langgraph.graph import StateGraph
```

```python
class AgentSchema(TypedDict): # state
    message:str
    
def greeting(state: AgentSchema) -> AgentSchema: # function (which is used to create the node)
    state['message'] = f"Hi {state['message']}, Say Hello World!!!"
    return state
```

```python
# creation of the node

graph = StateGraph(AgentSchema)

graph.set_entry_point("hello")
graph.add_node("hello",greeting)
graph.set_finish_point("hello")

compiled_graph = graph.compile()
```

```python
result = compiled_graph.invoke({"message":"Bro"})
```
